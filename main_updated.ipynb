```python
import random
from datetime import datetime, timedelta
import mysql.connector
```

```python
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=os.environ["MYSQL_PASSWORD"],
    database="zero_waste_logistics"
)
cursor = conn.cursor()

cursor.execute("SELECT node_id, amenity_type FROM location")
locations = cursor.fetchall() 

# 3. Setup Simulation Parameters
num_days = 30
end_date = datetime.now()
start_date = end_date - timedelta(days=num_days)

synthetic_data = []

print("Simulating historical requests...")
for node_id, amenity in locations:
    
    if amenity == 'restaurant':
        num_requests = random.randint(15, 45) 
    elif amenity == 'hospital':
        num_requests = random.randint(30, 60) 
    else:
        num_requests = random.randint(5, 20) 

    for _ in range(num_requests):
        random_seconds = random.randint(0, int((end_date - start_date).total_seconds()))
        request_time = start_date + timedelta(seconds=random_seconds)
        
        if amenity == 'restaurant' and random.random() < 0.6:
            request_time = request_time.replace(hour=random.randint(20, 23))

        volume = max(2.0, round(random.gauss(15.0, 5.0), 2))

        synthetic_data.append((node_id, request_time.strftime('%Y-%m-%d %H:%M:%S'), volume))

print(f"Generated {len(synthetic_data)} synthetic requests!")

insert_query = """
    INSERT INTO historical_requests (node_id, request_timestamp, waste_volume_kg)
    VALUES (%s, %s, %s)
"""

print("Pushing data to MySQL...")
cursor.executemany(insert_query, synthetic_data)
conn.commit()

cursor.close()
conn.close()

print("Data successfully loaded into the database!")
```

```python
from scipy.spatial import KDTree

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=os.environ["MYSQL_PASSWORD"], 
    database="zero_waste_logistics"
)
cursor = conn.cursor()

cursor.execute("SELECT node_id, name, latitude, longitude, amenity_type FROM location")
rows = cursor.fetchall()

locations_df = pd.DataFrame(rows, columns=['node_id', 'name', 'latitude', 'longitude', 'amenity_type'])

coordinates = locations_df[['latitude', 'longitude']].values

spatial_tree = KDTree(coordinates)

print(f"Success! Built a k-d tree containing {spatial_tree.n} spatial nodes.")

cursor.close()
conn.close()
```

```python
import os
import random

def get_truck_location(source="simulate", locations_df=locations_df):
    if source == "mysql":
        import mysql.connector
        conn = mysql.connector.connect(
            host="localhost",
            user="root",
            password=os.environ["MYSQL_PASSWORD"],
            database="zero_waste_logistics",
        )
        cursor = conn.cursor()
        cursor.execute("SELECT latitude, longitude FROM truck_telemetry ORDER BY timestamp DESC LIMIT 1")
        row = cursor.fetchone()
        conn.close()
        if row is None:
            raise ValueError("No telemetry rows found in truck_telemetry table.")
        return [row[0], row[1]]

    lat_min, lat_max = float(locations_df["latitude"].min()), float(locations_df["latitude"].max())
    lon_min, lon_max = float(locations_df["longitude"].min()), float(locations_df["longitude"].max())
    return [random.uniform(lat_min, lat_max), random.uniform(lon_min, lon_max)]

truck_location = get_truck_location(source="simulate")

distances, indices = spatial_tree.query(truck_location, k=5)

print("TRUCK CURRENT LOCATION:", truck_location)
print("-" * 50)
print("TOP 5 CLOSEST PICKUP LOCATIONS:")
print("-" * 50)

for rank, index in enumerate(indices):
    location_name = locations_df.iloc[index]['name']
    location_type = locations_df.iloc[index]['amenity_type']
    
    relative_dist = round(distances[rank], 4)
    
    print(f"{rank + 1}. {location_name} ({location_type}) - Distance Score: {relative_dist}")
```

```python
from scipy.spatial import distance_matrix

route_coordinates = [truck_location]
route_names = ["Garbage Truck (Start)"]

for index in indices:
    lat = locations_df.iloc[index]['latitude']
    lon = locations_df.iloc[index]['longitude']
    name = locations_df.iloc[index]['name']
    
    route_coordinates.append([lat, lon])
    route_names.append(name)
    
route_coordinates = np.array(route_coordinates, dtype = 'float')

dist_matrix = distance_matrix(route_coordinates, route_coordinates)

print(f"Successfully generated a {dist_matrix.shape[0]}x{dist_matrix.shape[1]} Distance Matrix.")
print("-" * 50)
print("Nodes in the routing network:")
for i, name in enumerate(route_names):
    print(f"Node {i}: {name}")
```

## Upgrade: Real Road-Network Distances

```python
import networkx as nx
import osmnx as ox
import os

GRAPH_PATH = "prayagraj_drive_network.graphml"

if os.path.exists(GRAPH_PATH):
    road_graph = ox.load_graphml(GRAPH_PATH)
    print(f"Loaded cached road graph from {GRAPH_PATH}")
else:
    print("Downloading road network from OpenStreetMap (this can take a minute)...")
    road_graph = ox.graph_from_place(place, network_type="drive")
    ox.save_graphml(road_graph, GRAPH_PATH)
    print(f"Downloaded and cached to {GRAPH_PATH}")

print(f"Road graph: {len(road_graph.nodes):,} nodes, {len(road_graph.edges):,} edges")
```

```python
lats = route_coordinates[:, 0]
lons = route_coordinates[:, 1]

snapped_node_ids = ox.distance.nearest_nodes(road_graph, X=lons, Y=lats)

print("Stop -> nearest road node:")
for name, node_id in zip(route_names, snapped_node_ids):
    print(f"  {name}: {node_id}")
```

```python
n = len(snapped_node_ids)
road_dist_matrix_m = np.zeros((n, n))  # distances in meters

for i, src_node in enumerate(snapped_node_ids):
    lengths = nx.single_source_dijkstra_path_length(road_graph, src_node, weight="length")
    for j, dst_node in enumerate(snapped_node_ids):
        if i == j:
            continue
        road_dist_matrix_m[i, j] = lengths.get(dst_node, np.inf)

finite_vals = road_dist_matrix_m[np.isfinite(road_dist_matrix_m)]
if finite_vals.size:
    penalty = finite_vals.max() * 10
    road_dist_matrix_m[~np.isfinite(road_dist_matrix_m)] = penalty

road_dist_matrix_km = road_dist_matrix_m / 1000.0

print("Real road-network distance matrix (km):")
road_dist_matrix_km
```

```python
euclidean_dist_matrix = dist_matrix.copy()
dist_matrix = road_dist_matrix_km

print("dist_matrix now contains real road-network distances (km).")
print(f"Euclidean total (sum of all pairs): {euclidean_dist_matrix.sum():.2f}")
print(f"Road-network total (sum of all pairs): {dist_matrix.sum():.2f} km")
```

```python
import numpy as np
import random

# 1. ACO Hyperparameters
num_ants = 10           # Number of artificial ants exploring per iteration
num_iterations = 50     # How many times the swarm will explore
evaporation_rate = 0.5  # How quickly bad trails disappear (50%)
alpha = 1.0             # Importance of pheromones (trail strength)
beta = 2.0              # Importance of visibility (distance heuristic)

num_nodes = dist_matrix.shape[0]

# 2. Initialize Pheromone Matrix (Starts with equal pheromones everywhere)
pheromones = np.ones((num_nodes, num_nodes))

# Initialize Visibility/Heuristic Matrix (Inverse of distance. Add 1e-10 to avoid dividing by zero)
heuristic = 1.0 / (dist_matrix + 1e-10)

best_path = None
best_path_length = float('inf')

print("Unleashing the ant colony...")

# 3. The Core ACO Loop
for iteration in range(num_iterations):
    paths = []
    path_lengths = []

    for ant in range(num_ants):
        # All ants start at the Garbage Truck (Node 0)
        current_node = 0
        path = [current_node]
        visited = set(path)
        path_length = 0

        # Build the path until all locations are visited
        while len(visited) < num_nodes:
            probabilities = np.zeros(num_nodes)

            # Calculate the probability of moving to each unvisited node
            for next_node in range(num_nodes):
                if next_node not in visited:
                    # The ACO Formula: Pheromone^alpha * Visibility^beta
                    tau = pheromones[current_node][next_node] ** alpha
                    eta = heuristic[current_node][next_node] ** beta
                    probabilities[next_node] = tau * eta
                # else stays 0.0 (already visited)

            # Normalize probabilities so they add up to exactly 1.0.
            # Using numpy division (instead of a Python list comprehension) keeps
            # this numerically stable, so np.random.choice below won't intermittently
            # throw "probabilities do not sum to 1" as num_nodes grows or when using
            # real road-network distances (which have a much wider value spread).
            prob_sum = probabilities.sum()
            if prob_sum > 0:
                probabilities = probabilities / prob_sum
            else:
                # Extremely unlikely edge case (all remaining options scored 0) --
                # fall back to a uniform choice among unvisited nodes so the ant
                # doesn't crash, it just picks something reasonable.
                unvisited = [n for n in range(num_nodes) if n not in visited]
                probabilities = np.zeros(num_nodes)
                probabilities[unvisited] = 1.0 / len(unvisited)

            # The ant randomly picks the next node, weighted by the probabilities
            next_node = np.random.choice(range(num_nodes), p=probabilities)

            # Update path and distance
            path.append(next_node)
            visited.add(next_node)
            path_length += dist_matrix[current_node][next_node]
            current_node = next_node

        # Optional: Have the truck return to the start (Closed-loop TSP)
        # path_length += dist_matrix[current_node][0]
        # path.append(0)

        paths.append(path)
        path_lengths.append(path_length)

        # Check if this ant found a new global best path
        if path_length < best_path_length:
            best_path = path
            best_path_length = path_length

    # 4. Pheromone Update Phase
    # First, evaporate old pheromones
    pheromones = pheromones * (1 - evaporation_rate)

    # Then, let the current ants deposit new pheromones based on path quality
    for i, path in enumerate(paths):
        for j in range(len(path) - 1):
            node_a = path[j]
            node_b = path[j + 1]
            # Shorter paths result in a higher pheromone deposit
            deposit = 1.0 / path_lengths[i]
            pheromones[node_a][node_b] += deposit
            pheromones[node_b][node_a] += deposit  # Because the route goes both ways

# 5. Output the Results
print("-" * 50)
print("OPTIMAL ROUTE FOUND:")
print("-" * 50)
for step, node_index in enumerate(best_path):
    print(f"Stop {step + 1}: {route_names[node_index]}")

print(f"\nTotal Route Distance Score: {round(best_path_length, 6)}")
```
